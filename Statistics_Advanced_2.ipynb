Q1: What is hypothesis testing in statistics?

Answer:

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population based on sample data. It involves formulating two hypotheses: the null hypothesis (H₀) and the alternative hypothesis (H₁).

The process checks whether the observed data provides enough evidence to reject the null hypothesis. Researchers use statistical tests and probability values to determine whether the results are statistically significant.

For example, a company may test whether a new product increases average sales compared to the old product.

Q2: What is the null hypothesis, and how does it differ from the alternative hypothesis?

Answer:
The Null Hypothesis (H₀) states that there is no significant effect or difference between variables. It represents the default assumption that nothing has changed.

The Alternative Hypothesis (H₁) states that there is a significant effect or difference.

Example
suppose a company claims that the average battery life of a phone is 10 hours.

 * Null Hypothesis (H₀): The average battery life = 10 hours

 * Alternative Hypothesis (H₁): The average battery life ≠ 10 hours

Key Difference
Null Hypothesis	          Alternative Hypothesis
Assumes no change	        Assumes change exists
Default assumption	      Opposite of H₀
Tested statistically	    Accepted if H₀ is rejected

Q3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

Answer:
The significance level (α) is the probability of rejecting the null hypothesis when it is actually true. It represents the risk of making a Type I error.

Common significance levels are:

 * 0.05 (5%)

 * 0.01 (1%)

 * 0.10 (10%)

Role in Decision Making

(1) Calculate the p-value from the statistical test.

(2) Compare it with the significance level (α).

  * If p-value ≤ α, reject the null hypothesis.

  * If p-value > α, fail to reject the null hypothesis.

For example, if α = 0.05 and p-value = 0.03, the null hypothesis is rejected because the result is statistically significant.

Q4: What are Type I and Type II errors? Give examples of each.

Answer:
In hypothesis testing, errors may occur when making decisions about hypotheses.

Type I Error

A Type I error occurs when the null hypothesis is rejected even though it is true.

Example:
A medical test says a person has a disease when they actually do not.

Type II Error

A Type II error occurs when the null hypothesis is not rejected even though it is false.

Example:
A medical test says a person is healthy when they actually have the disease.

Summary
Error Type	                 Meaning
Type I Error	          Rejecting a true null hypothesis
Type II Error	          Accepting a false null hypothesis


Q5: Difference between Z-test and T-test

Answer:
The Z-test and T-test are statistical tests used to compare sample means.

Z-Test	T-Test
Used when population standard deviation is known	Used when population standard deviation is unknown
Used for large sample size (n > 30)	Used for small sample size (n < 30)
Based on normal distribution	Based on t-distribution
Example

 * Use Z-test for large datasets like national survey data.

 * Use T-test for small datasets like classroom test scores.

```python
#Q7 Implement hypothesis testing using Z-statistics for a sample dataset in Python.Show the Python code and interpret the results.

import numpy as np
from scipy import stats

# Given sample data
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Hypothesized population mean
population_mean = 50

# Calculate sample statistics
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)

# Calculate Z statistic
z_score = (sample_mean - population_mean) / (sample_std / np.sqrt(n))

# Calculate p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

print("Sample Mean:", sample_mean)
print("Z-score:", z_score)
print("P-value:", p_value)

# Significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
#Q8: Write a Python script to simulate data from a normal distribution and calcalate the 95% confidence interval for its mean. Plot the data using Matplotlib.

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Step 1: Simulate data from a normal distribution
np.random.seed(0)   # for reproducibility
data = np.random.normal(loc=50, scale=5, size=100)

# Step 2: Calculate mean and standard error
mean = np.mean(data)
std_dev = np.std(data, ddof=1)
n = len(data)

# Step 3: Calculate 95% confidence interval
confidence_level = 0.95
z_value = stats.norm.ppf((1 + confidence_level) / 2)

margin_error = z_value * (std_dev / np.sqrt(n))

lower_bound = mean - margin_error
upper_bound = mean + margin_error

print("Sample Mean:", mean)
print("95% Confidence Interval:", (lower_bound, upper_bound))

# Step 4: Plot histogram
plt.hist(data, bins=15, edgecolor='black')
plt.title("Histogram of Simulated Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
#Q9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.


import numpy as np
import matplotlib.pyplot as plt

# Function to calculate Z-scores
def calculate_z_scores(data):
    mean = np.mean(data)
    std = np.std(data)

    z_scores = [(x - mean) / std for x in data]
    return z_scores

# Example dataset
data = [10, 12, 14, 15, 18, 20, 22, 25, 30]

# Calculate Z-scores
z_scores = calculate_z_scores(data)

print("Z-Scores:", z_scores)

# Plot histogram of standardized data
plt.hist(z_scores, bins=8, edgecolor='black')
plt.title("Histogram of Z-Scores")
plt.xlabel("Z-score")
plt.ylabel("Frequency")

plt.show()
```
