Q1.What is a random variable in probability theory?
   > A random variable is a variable that takes numerical values based on the outcome of a random experiment.
   🔹 Types of Random Variables
  (1) Discrete Random Variable

     * Takes countable values (finite or countably infinite).
     * Example:
        * Number of heads in 3 coin tosses → values {0, 1, 2, 3}.
        * Number of students absent in a class.

  (2) Continuous Random Variable

     * Takes uncountable values over an interval.
     * Example:
       * Time taken to complete a task (could be any real number between 0 and ∞)
       * Height or weight of a person.

Q2. What are the types of random variables?     
   > There are two main types of random variables in probability theory:
  1. Discrete Random Variable
   
   * Takes countable values (finite or countably infinite).
   * Each value has a specific probability.
   * Represented by a Probability Mass Function (PMF).

 Examples:
  * Number of heads in 3 coin tosses → {0, 1, 2, 3}
  * Number of students absent in a class
  * Result of rolling a die → {1, 2, 3, 4, 5, 6}

 2. Continuous Random Variable

  * Takes uncountably infinite values within an interval of real numbers.
  * Described by a Probability Density Function (PDF), not by individual probabilities.
  * The probability of taking any exact value is 0; only intervals have nonzero probability.

Examples:
   * Height or weight of a person
   * Time taken to finish a race
   * Temperature at noon in a city  


Q3. Explain the difference between discrete and continuous   distributions.?
 > Discrete Probability Distribution
   * Deals with discrete random variables (countable values).
   * Probabilities are assigned to specific values.
   * Represented by a Probability Mass Function (PMF).

   Examples:
   * Probability of getting 2 heads in 3 coin tosses.
   * Probability that a die shows 4

 * Continuous Probability Distribution
   * Deals with continuous random variables (uncountably infinite values).
   * Probabilities are assigned to intervals, not specific points.
   * Represented by a Probability Density Function (PDF).

   Examples:
   * Probability that a person’s height is between 160 cm and 170 cm.
   * Probability that temperature lies between 20°C and 25°C


Q4.What is a binomial distribution, and how is it used in probability?
  > A binomial distribution is a probability distribution that describes the number of successes in a fixed number of independent trials, where each trial has only two possible outcomes — success or failure.
  Key Characteristics:
   * There are a fixed number of trials (n).

   * Each trial has only two outcomes — success (with probability p) and failure (with probability 1 − p).

   * The trials are independent.

   * The probability of success (p) remains constant in each trial.

> Uses in Probability:

  * Modeling yes/no or success/failure events (e.g., defective vs. non-defective items).

  * Calculating probabilities in survey sampling, quality control, or clinical trials.

  * Used to approximate other distributions (e.g., normal distribution when n is large).

Q5. What is the standard normal distribution, and why is it important?
  > The standard normal distribution is a special type of normal distribution with:
  * Mean (μ) = 0

  * Standard deviation (σ) = 1

* Key Features:

  * It is symmetrical around 0.

  * The total area under the curve equals 1 (representing total probability).

  * The probability of outcomes can be found using Z-scores.
> Importance:-

 * It allows comparison between different normal distributions.

 * Probability tables (Z-tables) are based on it, helping find areas or probabilities.
	​
 * It is used widely in hypothesis testing, confidence intervals, and statistical inference.

 Q6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
  > The Central Limit Theorem (CLT) states that when a large number of independent, random samples are taken from any population with a finite mean (μ) and standard deviation (σ), the sampling distribution of the sample mean will approximate a normal distribution, regardless of the original population's shape.
> Key Points:
  * Works when sample size n is large (typically n ≥ 30)

  * The mean of the sampling distribution = population mean (μ).

  * The standard deviation of the sampling distribution (called standard error)

> Importance:

 * Allows use of normal distribution for inference even when data are not normally distributed.

 * Forms the basis for confidence intervals, hypothesis testing, and many statistical models.

 * Enables prediction and decision-making using sample data instead of the whole population.

Q7. What is the significance of confidence intervals in statistical analysis?
  > A confidence interval (CI) provides a range of values that is likely to contain the true population parameter (such as the mean or proportion) with a certain level of confidence.
* Key Idea:
  * It doesn’t give a single estimate but a range that reflects the uncertainty due to sampling.
* Significance:
  * Indicates how precise an estimate is — narrower intervals mean more precision.

  * Provides a measure of reliability for sample estimates.

  * Commonly used in research, quality control, and decision-making to assess uncertainty.

* Significance:

 * Helps in decision-making under uncertainty.

```python
#Q9. Write a Python program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.



import numpy as np
import matplotlib.pyplot as plt


mean = 50
std_dev = 5
data = np.random.normal(mean, std_dev, 1000)


calculated_mean = np.mean(data)
calculated_std = np.std(data)


print("Calculated Mean:", calculated_mean)
print("Calculated Standard Deviation:", calculated_std)


plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram of Normally Distributed Random Numbers')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.axvline(calculated_mean, color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.legend()
plt.show()
```

```python
#Q10: You are working as a data analyst for a retail company. The company has collected daily sales data for 2 years and wants you to identify the overall sales trend.
#daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255, 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
#● Explain how you would apply the Central Limit Theorem to estimate the average sales with a 95% confidence interval.
#● Write the Python code to compute the mean sales and its confidence interval.

 #The Central Limit Theorem (CLT) states that if we take many samples from a population, the sampling distribution of the sample mean will approximate a normal distribution, regardless of the population’s shape — provided the sample size is large enough (n ≥ 30 is a common rule of thumb).
#Even though our dataset is small (20 days), we can still apply the concept of CLT to estimate the population mean and build a 95% confidence interval for the average daily sales. Steps:-

#1. Compute the sample mean and sample standard deviation using NumPy.
#2. Use the formula for a 95% confidence interval:


import numpy as np
import scipy.stats as stats


daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]


sales = np.array(daily_sales)


mean_sales = np.mean(sales)
std_sales = np.std(sales, ddof=1)  # ddof=1 for sample std deviation
n = len(sales)


confidence_level = 0.95
z = stats.norm.ppf(1 - (1 - confidence_level)/2)
margin_of_error = z * (std_sales / np.sqrt(n))
ci_lower = mean_sales - margin_of_error
ci_upper = mean_sales + margin_of_error


print("Sample Mean Sales:", round(mean_sales, 2))
print("Sample Standard Deviation:", round(std_sales, 2))
print(f"95% Confidence Interval: ({round(ci_lower, 2)}, {round(ci_upper, 2)})")

```
